```python
class Solution(object):
    def singleNumber(self, nums):
        freq = {}
        for num in nums:
            if num in freq:
                freq[num] += 1
            else:
                freq[num] = 1
        for num in freq:
            if freq[num] == 1:
                return num
```

However, the question specifically says:

Linear time: O(n)

Constant extra space: O(1)

Our frequency dictionary uses O(n) extra space, so it does not satisfy the requirement.

The intended solution uses **XOR (^)**:

```python
def single_number(a):
  ans = 0
  for num in a:
    ans ^= num
  return ans
single_number([3,7,1,3,4,4,7])
```

Why XOR works:

2 ^ 2 = 0

1 ^ 1 = 0

0 ^ x = x

Your code:

ans = 0


---


for num in a:

    ans ^= num


---



is actually computing:

0 ^ 3 ^ 7 ^ 1 ^ 3 ^ 4 ^ 4 ^ 7

Now use the XOR properties:

x ^ x = 0

0 ^ x = x

Since **XOR is associative and commutative, we can rearrange**:

0 ^ 3 ^ 7 ^ 1 ^ 3 ^ 4 ^ 4 ^ 7

= 0 ^ (3 ^ 3) ^ (7 ^ 7) ^ (4 ^ 4) ^ 1

= 0 ^ 0 ^ 0 ^ 0 ^ 1

= 1

That's why the final answer is 1.
