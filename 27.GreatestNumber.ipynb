```python
# Method - 1
class Solution(object):
    def nextGreatestLetter(self, letters, target):
        for ch in letters:
            if ch > target:
                return ch
        return letters[0]
```

```python
# Method -2
class Solution(object):
    def nextGreatestLetter(self, letters, target):
        start = 0
        end = len(letters) - 1
        ans = letters[0]
        while start <= end:
            mid = (start + end)//2
            if letters[mid] > target:
                ans = letters[mid]
                end = mid - 1
            else:    # letters[mid] < target
                start = mid + 1
        return ans
```
