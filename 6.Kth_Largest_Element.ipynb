```python
# it gives TLE
class Solution(object):
    def findKthLargest(self, nums, k):
        for i in range(k):
            max_val = float("-inf")
            for j in range(len(nums)):
                if nums[j] > max_val:
                    max_val = nums[j]
            nums.remove(max_val)
        return max_val
```

1. Find the maximum value.
2. Find its index.
3. Put it at the end (its correct position).
4. Repeat for the remaining array.

Then that's essentially **Selection Sort** (selecting the maximum each time).

```python
# it also gives TLE -
class Solution(object):
    def findKthLargest(self, nums, k):
        for end in range(len(nums)-1,0,-1):
            max_index = 0
            for i in range(1, end+1):
                if nums[i] > nums[max_index]:
                    max_index = i
            nums[max_index], nums[end] = nums[end], nums[max_index]
        return nums[-k]
```

However...

This solution is basically Selection Sort.

Complexity: O(n²)

LeetCode accepts it for small inputs, but for large inputs it may give TLE (Time Limit Exceeded).

For this question, the expected approaches are:

**1. Min Heap → O(n log k)**

**2. Quick Select → Average O(n)**

Quick Sort recursively sorts both sides of the pivot.
Quick Select only recurses into the side that contains the k-th element.

**That's why Quick Select is faster when you only need one rank (like the 5th smallest element) instead of a fully sorted array.**
