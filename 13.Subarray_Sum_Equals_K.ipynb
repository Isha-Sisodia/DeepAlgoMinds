```python
# Brute Force approach -
# TLE

# a = [1,2,3]
# k = 3
a = [2, -1, 3, -3]
k = 2
n = len(a)
count = 0
for i in range(n):
  curr_sum = 0
  for j in range(i,n):
    curr_sum += a[j]
    if curr_sum == k:
      count += 1
print(count)
```

```python
# TLE
class Solution(object):
    def subarraySum(self, nums, k):
        n = len(nums)
        count = 0
        for i in range(n):
            curr_sum = 0
            for j in range(i,n):
                curr_sum += nums[j]
                if curr_sum == k:
                    count+= 1
        return count
```

###Optimized Approach -

 ### **(Prefix Sum + Hash Map)**

Idea -

Suppose at some index:

prefix_sum = 6

and

k = 3

If we have previously seen:

prefix_sum = 3

then:

6(Current_prefix_sum) - 3 (k) = 3(Earlier Prefix_sum)

which means the subarray between those positions (means between earlier and current prefix sum) has sum k.

So for every current prefix sum, we check:

prefix_sum - k

in a hash map.

```python
# Optimised Solutio using hasing and prefix sum
a = [2, -1, 3, -3]
k = 2
prefix_sum = 0
freq = {0:1} #freq stores: How many times each prefix sum has appeared so far.
count = 0
for num in a:
  prefix_sum += num
  need = prefix_sum - k

  if need in freq:
    count +=freq[need]

  if prefix_sum in freq:
    freq[prefix_sum] += 1
  else:
    freq[prefix_sum] = 1
print(count)
```

✅ Prefix sum = running sum

✅ We store previous prefix sums in freq

✅ {0:1} helps count subarrays starting from index 0

✅ We search for prefix_sum - k

Subarray Sum Equals K — Simple Revision Idea
Step 1: Keep a running sum
prefix_sum

This tells me:

"How much sum have I collected till now?"

Step 2: Current prefix sum = What I have

Suppose:

prefix_sum = P

and I want:

subarray_sum = k
Step 3: What earlier prefix sum do I need?

Ask:

P - Earlier_Prefix = k

Therefore:

Earlier_Prefix = P - k

So I search for:

prefix_sum - k
Step 4: If that earlier prefix exists

Then:

Current Prefix - Earlier Prefix = k

which means:

A valid subarray exists.

Step 5: If it exists multiple times

Suppose:

freq[prefix_sum - k] = 3

Then:

There are 3 different earlier positions.

So:

3 different valid subarrays exist.

That's why:

count += freq[prefix_sum - k]
Step 6: Why {0:1} ?

To count subarrays starting from index 0.

Example:

[2]
k = 2

Current prefix:

2

Need:

2 - 2 = 0

So we need one prefix sum 0 already present.

Hence:

freq = {0:1}
One-Line Memory Trick

👉 Current Prefix = What I have

👉 k = What I want

👉 Current Prefix − k = Earlier Prefix I need

👉 How many times that Earlier Prefix appeared = Number of valid subarrays
