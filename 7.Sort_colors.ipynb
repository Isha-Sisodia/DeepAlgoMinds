```python
# Method 1 - Count Frequency - based on Hashing idea..but not exactly hasing because we didnt use dictionary in this question
arr = [2,1,0,0,2,1,1,0,2]
count0 = count1 = count2 = 0
for num in arr:
    if num == 0:
        count0 += 1
    elif num == 1:
        count1 += 1
    else:
        count2 += 1
idx = 0
for i in range(count0):
    arr[i] = 0
    idx += 1
for i in range(count1):
    arr[idx] = 1
    idx += 1
for i in range(count2):
    arr[idx] = 2
    idx += 1
print(arr)
```

```python
# Method 2 :Dutch National Flag Algorithm
a = [2,1,0,0,2,1,1,0,2]
n = len(a)

start = 0
mid = 0
end = n - 1

while mid <= end:

    if a[mid] == 0:
        a[start], a[mid] = a[mid], a[start]
        start += 1
        mid += 1

    elif a[mid] == 1:
        mid += 1

    else:  # a[mid] == 2
        a[mid], a[end] = a[end], a[mid]
        end -= 1

print(a)

# we can use this approach here because array just can have 0,1,2 elements
# Time Compexity - O(n)
```

```python
# Method 3: Bubble Sort (Easy to understand)
nums = [2,0,2,1,1,0]

for i in range(len(nums)):
    for j in range(len(nums)-1):
        if nums[j] > nums[j+1]:
            nums[j], nums[j+1] = nums[j+1], nums[j]

print(nums)
```

```python
# Bubble Sort
# i submitted this
class Solution(object):
    def sortColors(self, nums):
        for i in range(len(nums)):
            for j in range(len(nums)-1):
                if nums[j] > nums[j+1]:
                    nums[j], nums[j+1] = nums[j+1], nums[j]
        return nums
```

```python
# Method 4: Selection Sort - also done in previouus question....difference is in this we use minimum here and in that question we used maximum

# Find the smallest element and place it at the beginning.

nums = [2,0,2,1,1,0]

for i in range(len(nums)):
    min_idx = i

    for j in range(i+1, len(nums)):
        if nums[j] < nums[min_idx]:
            min_idx = j

    nums[i], nums[min_idx] = nums[min_idx], nums[i]

print(nums)
```
