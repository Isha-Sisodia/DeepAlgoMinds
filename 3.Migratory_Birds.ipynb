## Migratory Birds — Question Explanation

You are given an array of integers.

Your task is to find:

The number that appears most frequently in the array.**(this means we have to use dictionary)**

If two or more numbers have the same highest frequency, return the smallest number among them.

### **Example 1**

Input: 1 4 4 4 5 3

Frequency of each number:

1 → 1 time

3 → 1 time

4 → 3 times

5 → 1 time

The number 4 appears the most (3 times).

Output: 4

### **Example 2**

Input: 1 2 3 4 5 4 3 2 1 3 4

Frequency of each number:

1 → 2 times

2 → 2 times

3 → 3 times

4 → 3 times

5 → 1 time

Both 3 and 4 have the highest frequency (3 times).

Since we must return the smallest number among those with the highest frequency:

Output: 3

## Steps to Solve -
1. Count how many times each number appears.
2. Find the maximum frequency.
3. Find all numbers having that frequency.
4. Return the smallest among them.

```python
# there are 2 metods to count frequency -

# Method 1: Using get() (Recommended)

arr = [1, 2, 1, 3, 2, 1]
freq = {}
for num in arr:
    freq[num] = freq.get(num, 0) + 1
print(freq)
```

```python
# Method 2: Using if-else

arr = [1, 2, 1, 3, 2, 1]
freq = {}
for num in arr:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print(freq)
```

## Dictionary frequency Counting Solution -

```python
arr = [1, 2, 1, 3, 2, 1]

freq = {}
for num in arr:
    freq[num] = freq.get(num, 0) + 1
# print(freq)
max_freq = max(freq.values())

# now we have to identify that key which has max_freq...so we use loop
answer = float('inf')
for key in freq:
  if freq[key] == max_freq:
    answer = min(answer, key)
print(answer)
```

## Optimization -

Why can we optimize?

The problem says:

Each value is guaranteed to be 1, 2, 3, 4, or 5.

Since there are only 5 possible values, we don't need a dictionary.

We can use a list to store frequencies.

```python
# Try it later
# Optimized HackerRank Solution
def migratoryBirds(arr):
    count = [0] * 6

    for num in arr:
        count[num] += 1

    max_count = max(count)

    for i in range(1, 6):
        if count[i] == max_count:
            return i

arr = [1, 2, 1, 3, 2, 1]

print(migratoryBirds(arr))
```
