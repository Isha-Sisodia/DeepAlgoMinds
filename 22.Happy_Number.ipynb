````python
# my earlier code that i wrote
def isHappy(n):
  sum_digits = 0
  while n > 0:
    digit = n%10 #Last digit
    n = n//10 #digits remaining after last
    sum_digits += digit**2
    if sum_digits == 1:
      return True
    else:
      return False
# I wrote the code to calculate the sum of squares of digits once.
````

This calculates the sum for one number.

Now you need an outer loop that keeps repeating the whole process.

Pseudo-code:


```python
while n != 1:
    calculate sum of squares of digits
    n = sum_digits
```
But there is another problem.

For n = 2:

2 → 4

4 → 16

16 → 37

37 → 58

58 → 89

89 → 145

145 → 42

42 → 20

20 → 4

Now we're back at 4!

This will keep repeating forever.

**So we need to remember numbers we've already seen.**

Because:

- If loop stopped because n == 1 → return True
- If loop stopped because n in seen → return False

````python
n = 2

seen = set()

while n != 1 and n not in seen:
    seen.add(n)

    sum_digits = 0

    while n > 0:
        digit = n % 10  #last digit
        n = n // 10     # digits remaining after last
        sum_digits += digit ** 2

    n = sum_digits

print(n == 1)
````

**My Ques -** is it necessary that if a number not ends in 1 then it appears back?

**Answer** - yes, for this problem, if the sequence does not reach 1, it must eventually repeat a number.

Here's why.

When you repeatedly replace a number with the sum of the squares of its digits, the values cannot keep growing forever.

For example, suppose n = 999999999.

9² + 9² + 9² + 9² + 9² + 9² + 9² + 9² + 9²

= 81 × 9

= 729

A huge number quickly becomes much smaller.

After a few iterations, every number falls into a relatively small range.

Now think about it:

There are only finitely many numbers in that range.

If you never reach 1, but keep generating numbers forever,
then eventually you must revisit a number you've seen before.

This is the same idea as:

Possible numbers: 1, 2, 3, 4, 5

Sequence:
2 → 4 → 5 → 3 → 4 → 5 → 3 → ...

Once 4 appears again, the entire sequence repeats forever.
