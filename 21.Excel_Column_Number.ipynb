```python
def titleToNumber(columnTitle):
  ans = 0
  for ch in columnTitle:
    # value = ord('ch') - ord('A') + 1   #i got error beacuse i wrote ord("ch") instead of ord(ch)
    value = ord(ch) - ord("A") + 1
    ans = ans*26 + value
  return ans

titleToNumber("A")   #1
titleToNumber("AB")  #28
```
