Link - https://leetcode.com/problems/find-the-duplicate-number/description/

You are given an array nums of size n + 1, where every element is an integer in the range 1 to n.

Since there are n + 1 elements but only n possible values (1 to n), at least one number must be repeated.

Your task is to find and return the duplicate number.

Key Observation
- Array size = n + 1
- Allowed values = 1 to n
- More elements than available values ⇒ a duplicate is guaranteed to exist.

The question says:

- There is only one repeated number.
- This does not mean it appears exactly twice.

```python
# Approach 1 - Brute Force
a = [1,2,2,3,4]
for num in a:
  if a.count(num) > 1:
    print(num)
    break
# O(n²) - Time Complexity
# Worst
```

```python
# Approach 2
# Optimal Solution -Using Hasing

a = [1,2,2,3,4]
freq = {}
for num in a:
  if num in freq:
    freq[num] += 1
  else:
    freq[num] = 1
for num in freq:
  if freq[num] > 1:
    print(num)
    break
# O(n) - Time Complexity
# Good
```

```python
# Approach 3 - Hash Set -also optimal
a = [1,2,2,3,4]
seen = set()
for num in a:
  if num in seen:
    print(num)
  seen.add(num)
  # O(n) - Time Complexity
  # Best
```

```python
# Approach 4 - Sort
nums = [1,2,2,3,4]
nums.sort()

for i in range(1, len(nums)):
    if nums[i] == nums[i-1]:
        print(nums[i])
```
