**Key Observation**

If a number is ugly, we should be able to keep dividing by 2, 3, and 5 until we reach 1.

Example: 30

30 ÷ 2 = 15

15 ÷ 3 = 5

5 ÷ 5 = 1

Reached 1 → Ugly Number.

```python
def isUgly( n):
  if n <= 0:
    return False
  while n % 2 == 0:
    n = n//2
  while n % 3 == 0:
    n = n//3
  while n % 5 == 0:
    n = n//5
  return n == 1
isUgly(30)
```
