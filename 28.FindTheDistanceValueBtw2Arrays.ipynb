https://leetcode.com/problems/find-the-distance-value-between-two-arrays/

```python
class Solution(object):
    def findTheDistanceValue(self, arr1, arr2, d):
        count = 0
        for x in arr1:
            found = 0
            for y in arr2:
                if abs(x-y) <= d:
                    found = 1
                    break
            if found == 0:
                count += 1
        return count
```
