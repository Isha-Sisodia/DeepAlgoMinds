Question - Convert Binary to Decimal

```python
# Method 1 Left to Right (Using Powers)
n = "1011"
ans  = 0
for i in range(len(n)):
  ans += int(n[i])*(2**(len(n)-1-i))
print(ans)

# Method 2  Right to Left (Using Powers)
n = "1011"
ans = 0

for i in range(len(n)-1, -1, -1):
    ans += int(n[i]) * (2**(len(n)-1-i))

print(ans)

# Method 3: Without using ** (Most Efficient / Interview Favorite)
binary = "1011"

decimal = 0

for digit in binary:
    decimal = decimal * 2 + int(digit)

print(decimal)
```

Question - Counting Set bits

```python
# Method 1
n = 13
count = 0
while n > 0 :
  ans = n%2
  if ans == 1:
    count += 1
  n = n//2
print(count)

# or

n = 13
count = 0

while n > 0:
    count += n % 2
    n //= 2

print(count)
```

```python
# Method 2 - Method 2: Using & 1

n = 13
count = 0

while n > 0:
    count += (n & 1)
    n >>= 1         #so that jo count ho gya hai vo again count na ho !

print(count)
```

```python
# Faster Way (Brian Kernighan Algorithm)
# Instead of checking every bit:
n = 13
count = 0

while n:
    n = n & (n - 1)
    count += 1

print(count)
```

Counting Different Bits Between Two Numbers

```python
# Method - 1
a = 10    #10 = 01010
b = 20    #20 = 10100
count = 0
x = a ^ b   #now we just have to count set bits  (using %2)
while x:    #while x: means: -  while x != 0: or while x > 0 (for +ve integers):
  count += x%2
  x = x//2
print(count)

# Method - 2
a = 10    #10 = 01010
b = 20    #20 = 10100
count = 0
x = a ^ b   #now we just have to count set bits  (Using & and >>)
while x != 0:
  count += (x & 1)
  x >>= 1
print(count)
```

Given a number n and a position i, determine whether the ith bit of n is set (1) or unset (0).

What are we trying to find?

We want to answer:

"At position i, is the bit 1 or 0?"

If it's 1 → Set

If it's 0 → Unset

That's the entire goal of the question. 😊

For n = 13, i = 2:

Answer: The 2nd bit is set. ✅

```python
n = 13
i = 2

if n & (1 << i):
    print("Set")
else:
    print("Unset")
```
