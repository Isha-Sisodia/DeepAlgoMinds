Problem Link- https://leetcode.com/problems/rearrange-array-elements-by-sign/description/

Given -

nums = [3,1,-2,-5,2,-4]

Rule 1: Start with a positive number

Rule 2: Alternate signs

Positive → Negative → Positive → Negative ...

Rule 3: Preserve order

What the problem is really asking

**Step 1** - Separate positives and negatives.

positives = [3,1,2]

negatives = [-2,-5,-4]

**Step 2** - Create a new array.

**Step 3** - Take: [positive,negative,positive,negative
...]

until both lists are exhausted.

```python
# Submitted solution -
class Solution(object):
    def rearrangeArray(self, nums):
        positive = []
        negative = []

        for num in nums:
            if num > 0:
                positive.append(num)
            else:
                negative.append(num)

        ans = []

        for i in range(len(positive)):
            ans.append(positive[i])
            ans.append(negative[i])

        return ans
```

```python
# Method -2
positive = []
negative = []

for num in nums:
    if num > 0:
        positive.append(num)
    else:
        negative.append(num)

ans = [0] * len(nums)

p_index = 0
n_index = 0

for i in range(len(nums)):

    if i % 2 == 0:
        ans[i] = positive[p_index]
        p_index += 1
    else:
        ans[i] = negative[n_index]
        n_index += 1

return ans
```

Method 3 - Even Better

Instead of checking i % 2, we can directly place positives at even positions and negatives at odd positions.

```python
# Method 3 -

ans = [0] * len(nums)

pos = 0
neg = 1

for num in nums:
    if num > 0:
        ans[pos] = num
        pos += 2
    else:
        ans[neg] = num
        neg += 2

return ans
```
