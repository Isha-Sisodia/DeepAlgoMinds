```python
# Practice Question -
# Ques - Make a new array which contains only unique elements
a1 = [1,2,2,3,3,4,5]
result =[]
for num in a1:
  if num not in result:
    result.append(num)
print(result)
# but problem is time complexity is O(n^2)
# lets try to optimize
```

```python
# Method 1 but not optimized
class Solution(object):
    def intersection(self, nums1, nums2):
        result =[]
        for num in nums1:
            if num in nums2 and num not in result:
                result.append(num)
        return result
```

```python
# Using Hashing - Optimized Solution
# Better Solution
n1 = [2,3,1,4,5]
n2 = [4,5,1,9]
num1 = set(n1)
num2 = set(n2)

freq = {}
for num in num1:
  if num in freq:
    freq[num] +=1
  else:
    freq[num] = 1
for num in num2:
  if num in freq:
    freq[num] +=1
  else:
    freq[num] = 1

result = []

for num in freq:
  if freq[num] == 2:
    result.append(num)

# for key, value in freq.items():
#   if value == 2:
#     result.append(key)

print(result)
```
