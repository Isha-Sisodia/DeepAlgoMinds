```python
# make a new array which contains only unique elements
a1 = [1,2,2,3,3,4,5]
result =[]
for num in a1:
  if num not in result:
    result.append(num)
print(result)
# but problem is time complexity is O(n^2)
# lets try to optimize
```

My submission Link - https://leetcode.com/problems/intersection-of-two-arrays/


```python
# Optimized Solution - Using two Pointer
a = [1,2,2,3,3,4,5]
i = 0
j = 1
n = [a[0]]
for j in range(len(a)):
  if a[i] != a[j]:
    n.append(a[j])
    i = j
  j += 1
print(n)
```
