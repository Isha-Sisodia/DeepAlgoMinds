```Approach

For mid:

- If nums[mid] == target, return mid.

-- Check if the left half is sorted:

               nums[low] <= nums[mid]

- If target lies within this sorted half:

                nums[low] <= target && target < nums[mid]

search left.

- Else search right.
--  Otherwise, the right half is sorted:

- If target lies within it:

         nums[mid] < target && target <= nums[high]

search right.

- Else search left.

````python
class Solution(object):
    def search(self, nums, target):
        start = 0
        end = len(nums) - 1

        while start <= end:
            mid = (start + end) // 2

            if nums[mid] == target:
                return mid

            # Left half sorted
            if nums[start] <= nums[mid]:

                if nums[start] <= target < nums[mid]:
                    end = mid - 1
                else:
                    start = mid + 1

            # Right half sorted
            else:

                if nums[mid] < target <= nums[end]:
                    start = mid + 1
                else:
                    end = mid - 1

        return -1
````
