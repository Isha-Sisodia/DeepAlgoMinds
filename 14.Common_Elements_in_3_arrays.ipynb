```python
a = [1, 5, 10, 20, 40, 80]
b = [6, 7, 20, 80, 100]
c = [3, 4, 15, 20, 30, 70, 80, 120]

a1 = set(a)
b1 = set(b)
c1 = set(c)

result = []

freq = {}

for num in a1:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

for num in b1:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

for num in c1:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

for num in freq:
    if freq[num] == 3:
        result.append(num)

result.sort()   # To ensure non-decreasing order
# return sorted(result)
print(result)
```

we can also make it a little shorter using .get():

```python
a1 = set(a)
b1 = set(b)
c1 = set(c)

freq = {}

for num in a1:
    freq[num] = freq.get(num, 0) + 1

for num in b1:
    freq[num] = freq.get(num, 0) + 1

for num in c1:
    freq[num] = freq.get(num, 0) + 1

result = []

for num in freq:
    if freq[num] == 3:
        result.append(num)

result.sort()
print(result)
```

```python
# Method 2 in class -
# Uisng three Pointers

a = [1, 5, 16, 18, 20, 30]
b = [5, 16, 18, 20]
c = [5, 20]
common = []

i = j = k = 0
n1 = len(a)
n2 = len(b)
n3 = len(c)

while i < n1 and j < n2 and k < n3:
  if a[i] == b[j] and b[j] == c[k]:
    common.append(a[i])
    i += 1
    j += 1
    k += 1
  else:
    min_val = min(a[i], b[j], c[k])
    if a[i] == min_val:
      i += 1
    elif b[j] == min_val:
      j += 1
    else:
      k += 1

print(common)
```
