## Brute Force Method

Since the array is sorted, an efficient solution uses binary search, but if you want the brute force approach, simply traverse the entire array and record the first and last positions where x appears.

**Algorithm** -
1. Initialize first = -1 and last = -1.
2. Traverse the array from index 0 to n-1.
3. If arr[i] == x:
- If first == -1, set first = i.
- Set last = i.
4. After traversal, return {first, last}.

```python
# Brute force
class Solution:
    def find(self, arr, x):
        first = -1
        last = -1
        n = len(arr)
        for i in range(n):
            if arr[i] == x:
                if first == -1:
                    first = i
                last = i
        return([first, last])
```

```python
# Binary Search
class Solution:
    def find(self, arr, x):
        def first_occ():
            ans = -1
            start = 0
            end = len(arr) - 1
            while start <= end:
                mid = (start+end)//2
                if arr[mid] == x:
                    ans = mid
                    end = mid - 1
                elif arr[mid] > x:
                    end = mid - 1
                else:
                    start = mid + 1
            return ans
        def last_occ():
            ans = -1
            start = 0
            end = len(arr) - 1
            while start <= end:
                mid = (start+end)//2
                if arr[mid] == x:
                    ans = mid
                    start = mid + 1
                elif arr[mid] < x:
                    start = mid+1
                else:
                    end = mid - 1
            return ans
        return([first_occ(),last_occ()])
```

```python
# Using loop without two functions
class Solution(object):
    def searchRange(self, nums, target):

        first = -1
        last = -1

        # Find first occurrence
        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                first = mid
                right = mid - 1
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        # Find last occurrence
        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                last = mid
                left = mid + 1
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        return [first, last]
```
