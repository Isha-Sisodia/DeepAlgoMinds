```python
class Solution(object):
    def searchRange(self, nums, target):
        def first_occ():
            ans = -1
            start = 0
            end = len(nums) - 1
            while start <= end:
                mid = (start+end)//2
                if nums[mid] == target:
                    ans = mid
                    end = mid - 1
                elif nums[mid] > target:
                    end = mid - 1
                else: #nums[mid] < target:
                    start = mid + 1
            return ans
        def last_occ():
            ans = -1
            start = 0
            end = len(nums) - 1
            while start <= end:
                mid = (start+end)//2
                if nums[mid] == target:
                    ans = mid
                    start = mid + 1
                elif nums[mid] > target:
                    end = mid - 1
                else: # nums[mid] < target:
                    start = mid + 1
            return ans
        return([first_occ(), last_occ()])
```

```python
# Using loop without two functions
class Solution(object):
    def searchRange(self, nums, target):

        first = -1
        last = -1

        # Find first occurrence
        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                first = mid
                right = mid - 1
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        # Find last occurrence
        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                last = mid
                left = mid + 1
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        return [first, last]
```
