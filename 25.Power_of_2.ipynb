Before thinking about code, let's understand the pattern.

Powers of 2
2⁰ = 1

2¹ = 2

2² = 4

2³ = 8

2⁴ = 16

2⁵ = 32

2⁶ = 64
...

**If n is a power of 2, then we can keep dividing it by 2 and eventually reach 1.**

Example: 16

16 → 8 → 4 → 2 → 1

Example: 12

12 → 6 → 3

Now 3 is not divisible by 2, so 12 is not a power of 2.

```python
# TLE
n = 16

while n % 2 == 0:
    n = n // 2

print(n == 1)
```

**Dry Run -**

For n = 16

n = 16

16 % 2 == 0 → n = 8

8 % 2 == 0 → n = 4

4 % 2 == 0 → n = 2

2 % 2 == 0 → n = 1

return n == 1

Output: True

```python
# Accepted
n = 16

if n == 0:
  print(False)

while n % 2 == 0:
    n = n // 2

print(n == 1)
```

```python
# Submitted Code -
class Solution(object):
    def isPowerOfTwo(self, n):
        if n <= 0:
            return False
        while n % 2 == 0:
            n = n//2
        return n == 1
```
