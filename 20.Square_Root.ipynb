Observation -

Check some numbers:

- 0² = 0
- 1² = 1
- 2² = 4
- 3² = 9
- 4² = 16
- 5² = 25

Notice:

0, 1, 4, 9, 16, 25 ...

These square values are sorted.

That's the reason binary search can be used.

Your linear-search idea and binary search are actually doing the same thing:

Find the largest number i such that i*i <= x.

The difference is how they search.

Using Binary search -


```python
def mySqrt(x):
    start = 0
    end = x
    ans = 0

    while start <= end:
        mid = (start + end) // 2

        if mid * mid == x:
            return mid

        elif mid * mid < x:
            ans = mid
            start = mid + 1

        else:
            end = mid - 1

    return ans
```

Notice how ans plays the same role as i-1 in your solution.

In your code, when you first find i*i > x, you return i-1.
In binary search, whenever mid*mid < x, we store mid in ans because it is a valid candidate. If we never find a perfect square, ans becomes the floor of √x.

So binary search is not a different idea. It's the same idea (find the largest square ≤ x), just searching much faster.
