Instead, while moving left → right:

1. Keep the minimum price seen so far (possible buying day).
2. Treat the current price as the selling day.
3. Compute profit.

```python
prices = [7,1,5,3,6,4]
n = len(prices)
min_price = float("inf")
max_profit = float("-inf")
for i in range(n):
  min_price = min(min_price, prices[i])
  profit = prices[i] - min_price

  max_profit = max(max_profit, profit)
print(max_profit)
```
