 So the logic is:

while p1 >= 0 and p2 >= 0:

    compare and place larger element

while p2 >= 0:

    copy leftover elements of a2

```python
# Method 2 - Optimal Two Pointer Approach

a1 = [1,2,3,0,0,0]
n1 = 3

a2 = [2,5,6]
n2 = 3

p1 = n1 - 1
p2 = n2 - 1
k = n1 + n2 - 1

while p1 >= 0 and p2 >= 0:
    if a1[p1] > a2[p2]:
        a1[k] = a1[p1]
        p1 -= 1
    else:
        a1[k] = a2[p2]
        p2 -= 1

    k -= 1

while p2 >= 0:
    a1[k] = a2[p2]
    p2 -= 1
    k -= 1

print(a1)
```

```python
class Solution(object):
    def merge(self, nums1, m, nums2, n):
        p1 = m-1
        p2 = n-1
        k = m+n-1
        while p1 >= 0 and p2 >= 0:
            if nums1[p1] > nums2[p2]:
                nums1[k] = nums1[p1]
                p1 -= 1
            else:
                nums1[k] = nums2[p2]
                p2 -= 1
            k -= 1
        while p2 >= 0:
            nums1[k] = nums2[p2]
            k -= 1
            p2 -= 1
        return nums1

```
