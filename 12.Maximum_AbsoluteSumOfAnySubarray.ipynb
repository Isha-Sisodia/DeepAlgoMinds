We need:

Maximum Absolute Sum of Any Subarray

That means:

1. Generate all possible subarrays.
2. Find each subarray's sum.
3. Take absolute value. (Negative becomes positive.)
4. Return the largest one.

Example 1

nums = [1,-3,2,3,-4]

Some subarrays:

[1]        -> abs(1) = 1

[-3]       -> abs(-3) = 3

[2,3]      -> abs(5) = 5

[-3,2,3]   -> abs(2) = 2

[-4]       -> abs(-4) = 4

Maximum:

5

Answer:

5

```python
# Brute force approach
# a = [1,-3,2,3,-4]
a = [2,-5,1,-4,3,-2]
n = len(a)
max_sum = float("-inf")
for i in range(n):
  curr_sum = 0
  for j in range(i,n):
    curr_sum += a[j]
    max_sum = max(max_sum, abs(curr_sum))
print(max_sum)
```

```python
# Optimal Approach using kadanes algorithm two times
a = [2,-5,1,-4,3,-2]
n = len(a)

max_sum = float("-inf")
min_sum = float("inf")
curr_max = 0
curr_min = 0

for i in range(n):
  # Maximum subarray sum
  curr_max = max(a[i], curr_max + a[i])
  max_sum = max(max_sum, curr_max)

  # Minimum subarray sum
  curr_min = min(a[i], curr_min + a[i])
  min_sum = min(min_sum, curr_min)

print(max(max_sum, abs(min_sum)))
```

Key Idea

This problem is actually:

**Kadane's Algorithm × 2**

Find:

Maximum subarray sum
Minimum subarray sum

Then:

answer = max(
    abs(max_sum),
    abs(min_sum)
)

For:

nums = [2,-5,1,-4,3,-2]

Maximum subarray sum:

3

Minimum subarray sum:

-8

Absolute values:

abs(3) = 3
abs(-8) = 8

Answer:8
