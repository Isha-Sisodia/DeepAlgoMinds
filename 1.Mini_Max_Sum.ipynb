## Problem Description
Given - You are given 5 positive integers.

Example: 1 2 3 4 5

What do you need to do?

You must find:

Minimum sum possible by adding exactly 4 out of the 5 numbers.
Maximum sum possible by adding exactly 4 out of the 5 numbers.

## Method 1  - time complexity - O(n^2)

Let's use the example

Numbers:

1 2 3 4 5

Since we need the sum of 4 numbers, we leave out one number at a time.

Leave out 1

2 + 3 + 4 + 5 = 14

Leave out 2

1 + 3 + 4 + 5 = 13

Leave out 3

1 + 2 + 4 + 5 = 12

Leave out 4

1 + 2 + 3 + 5 = 11

Leave out 5

1 + 2 + 3 + 4 = 10

So the possible sums are:

14, 13, 12, 11, 10

Smallest sum = 10

Largest sum = 14

Output:

10 14

```python
def miniMaxSum(arr):
    sums = []

    for i in range(5):
        current_sum = 0

        for j in range(5):
            if i != j:      # Skip one number
                current_sum += arr[j]

        sums.append(current_sum)

    print(min(sums), max(sums))
arr = [1,2,3,4,5]
miniMaxSum(arr)
```

## Method 2 - Time Complexity - O(n)

Method 2: Use Total Sum (More Efficient)

First find the total sum.

1 + 2 + 3 + 4 + 5 = 15

Now:

Minimum sum = Total − Largest number

15 − 5 = 10

Maximum sum = Total − Smallest number

15 − 1 = 14

Output:

10 14

```python
def miniMaxSum(arr):
    total = sum(arr)

    min_sum = total - max(arr)
    max_sum = total - min(arr)

    print(min_sum, max_sum)


arr = [1,2,3,4,5]
miniMaxSum(arr)
```
