### Description
Rotate Array
Problem Statement

Given an integer array nums and a non-negative integer k, rotate the array to the right by k steps.

After each rotation:

The last element moves to the first position.
All other elements shift one position to the right.
Example 1

Input

nums = [1,2,3,4,5,6,7]
k = 3

Output

[5,6,7,1,2,3,4]

Explanation

After 1 rotation: [7,1,2,3,4,5,6]
After 2 rotations: [6,7,1,2,3,4,5]
After 3 rotations: [5,6,7,1,2,3,4]

this approach gives TLE
1. Save the last element.
2. Shift everything one position to the right.
3. Put the saved last element at index 0.
4. Repeat k times.

```python
nums = [1,2,3,4,5,6,7]
k = 3
for _ in range(k):
    last = nums[-1]

    for i in range(len(nums)-1, 0, -1):
        nums[i] = nums[i-1]

    nums[0] = last
print(nums)
```

```python
# Gives TLE
class Solution(object):
    def rotate(self, nums, k):
        for _ in range(k):
            last = nums[-1]
            for i in range(len(nums)-1,0,-1):
                nums[i] = nums[i-1]
            nums[0] = last
        return nums
```

Rotating 10 times is the same as:

7 rotations + 3 rotations

After 7 rotations, the array comes back to its original state.

So only 3 rotations are actually needed.

Revision Line

## Another Method to solve this question is -

The reverse (inverse) method is the standard true in-place solution for the Rotate Array problem.

Idea -
1. Reverse the entire array.
2. Reverse the first k elements.
3. Reverse the remaining n-k elements.

**Step 1:** Reverse whole array

[1,2,3,4,5,6,7]

↓

[7,6,5,4,3,2,1]

**Step 2:** Reverse first k=3 elements

[7,6,5,4,3,2,1]

↓

[5,6,7,4,3,2,1]

**Step 3:** Reverse remaining elements

[5,6,7,4,3,2,1]

↓

[5,6,7,1,2,3,4]

```python
def reverse(nums, left, right):
  while left < right:
    nums[left], nums[right] = nums[right], nums[left]
    left += 1
    right -= 1

def rotate(nums, k):
  n = len(nums)

  reverse(nums, 0, n-1)  # step 1 -reverse whole array
  reverse(nums, 0, k-1)  # step 2 - reverse first k elements
  reverse(nums, k, n-1)  # step 3 - reverse remaining
  return nums

rotate([1,2,3,4,5,6,7],3)
```
