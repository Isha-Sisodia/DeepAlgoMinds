````python
n = 5

if n == 1:
    print([[1]])
else:
    a = [[1], [1, 1]]
    prev = a[-1]

    for i in range(2, n):
        ans = [1]

        for j in range(len(prev) - 1):
            ans.append(prev[j] + prev[j + 1])

        ans.append(1)

        a.append(ans)
        prev = ans

    print(a)
````

````python
class Solution(object):
    def generate(self, numRows):
        if numRows == 1:
            return [[1]]
        else:
            a = [[1],[1,1]]
            prev = a[-1]
            for i in range(2,numRows):
                ans = [1]
                for j in range(len(prev)-1):
                    ans.append(prev[j] + prev[j+1])
                ans.append(1)
                a.append(ans)
                prev = ans
        return a
````
