**This is essentially what the merge step of Merge Sort does.**

the two arrays must already be sorted.

```python
# the core logic is this -
# the difference is just of the way of giving input to the function..lets write that code -
def merge(a1,a2):
  a1 = [4,5,6]
  a2 = [1,2,2,3]

  n1 = len(a1)
  n2 = len(a2)

  p1 = n1 - 1      #3
  p2 = n2 - 1      #3
  k = n1 + n2 - 1  #5

  ans = [0]*(k+1)   # we add 1 in k because k = 5 and we need 6 size array.

  while p1 >= 0 and p2 >= 0:
    if a1[p1] > a2[p2]:
        ans[k] = a1[p1]
        p1 -= 1
    else:
        ans[k] = a2[p2]
        p2 -= 1

    k -= 1

  while p2 >= 0:
      ans[k] = a2[p2]
      p2 -= 1
      k -= 1
  while p1 >= 0:
    ans[k] = a1[p1]
    p1 -= 1
    k -= 1
  return ans

merge([1,2,3],[2,4,5,6])
```

**Merge Sort is a classic Divide and Conquer algorithm.**

It works in 3 steps:

**1. Divide**

Split the array into two halves.
Example:

[5,2,4,1]

↓

[5,2]   [4,1]

↓

[5] [2] [4] [1]

**2. Conquer**

Recursively sort the smaller subarrays.

Single-element arrays are already sorted:

[5]
[2]
[4]
[1]

No work needed here.

**3. Combine (Merge)**

Merge the sorted halves:

[5] + [2]

↓

[2,5]

and

[4] + [1]

↓

[1,4]

Then:

[2,5] + [1,4]

↓

[1,2,4,5]


                [5,2,4,1]
               /         \
          [5,2]         [4,1]
          /   \         /   \
        [5]  [2]     [4]   [1]
          \   /         \   /
         [2,5]         [1,4]
              \       /
              [1,2,4,5]


Split → Split → Split → Merge → Merge → Merge

```python
def merge(a, start, mid, end):

    left = a[start:mid+1]
    right = a[mid+1:end+1]

    i = 0
    j = 0
    k = start

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            a[k] = left[i]
            i += 1
        else:
            a[k] = right[j]
            j += 1
        k += 1

    while i < len(left):
        a[k] = left[i]
        i += 1
        k += 1

    while j < len(right):
        a[k] = right[j]
        j += 1
        k += 1


def merge_sort(a, start, end):

    if start >= end:
        return

    mid = (start + end) // 2

    merge_sort(a, start, mid)     # Divide left half
    merge_sort(a, mid + 1, end)   # Divide right half

    merge(a, start, mid, end)     # Combine


a = [5,2,4,1]
merge_sort(a, 0, len(a)-1)

print(a)
```
