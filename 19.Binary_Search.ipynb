# Binary Search

### When is Binary Search Used?

Binary Search is used when the search space is **monotonic** (sorted in increasing/decreasing order or follows a monotonic condition).

Binary search is used with monotonic function (strictly increasing or strictly decreasing) and we have to search number...then we will use binary search

**Time Complexity:** O(log n)

---

### Case 1: Unsorted Array

```python
a = [2, 1, 4, 5, 8, 6]
k = 8
```

To find `k = 8`, we must scan the array element by element.

```python
for i in range(len(a)):
    if a[i] == k:
        print(i)
```

**Output:** `4`

**Time Complexity:** O(n)

Since the array is unsorted, Binary Search cannot be applied directly.

---

### Case 2: Sort First, Then Apply Binary Search

1. Sort the array → O(n log n)
2. Apply Binary Search → O(log n)

Total Complexity:

```text
O(n log n) + O(log n)
≈ O(n log n)
```

---

### Conclusion

For the given example, using a simple linear search (O(n)) is better than sorting the array and then applying Binary Search (O(n log n)), because the array is unsorted and we only need to search for one element.

Binary Search is most useful when:

* The data is already sorted, or
* Multiple search queries will be performed on the same sorted data.

````python
# Method 1 - we use mid = (start+end)//2
# But Sometimes in some questions (eg = start is 1 and end is 100) when we add 1+100...means number becomes larger....due to which sometimes question gives error according to constraint of size of integer
# And in this kind of problem we use method 2 where we dont add....we subtract

# This is Method -1
def search(nums, target):
    start = 0
    end = len(nums) - 1

    while start <= end:
        mid = (start + end) // 2

        if target < nums[mid]:
            end = mid - 1

        elif target > nums[mid]:
            start = mid + 1

        else:    #nums[mid] == target
            return mid

    return -1
search([-1,2,3,6,7],6)
````

````python
# Method 2....where we use subtraction
class Solution(object):
    def search(self, nums, target):
        start = 0
        end = len(nums) - 1
        while start <= end:
            mid = start + (end - start) // 2
            if target < nums[mid]:
                end = mid - 1
            elif target > nums[mid]:
                start = mid + 1
            else:    #nums[mid] == target
                return mid
        return -1
````

The second formula is used mainly to avoid integer overflow in languages like C++ and Java when start and end are very large. In Python, you can safely use either one, but in interviews and standard implementations you'll usually see:

For Binary Search, both formulas give the same mid value at every step.

Binary Search Logic
1. Array must be sorted.
2. Find the middle element.
3. Compare middle element with target.
4. If target equals middle → found.
5. If target is greater than middle → search in right half.
6. If target is smaller than middle → search in left half.
7. Repeat steps 2–6 on the remaining half.
8. If no elements are left to search → target not found.
