https://leetcode.com/problems/binary-search/

```python
def search(nums, target):
    start = 0
    end = len(nums) - 1

    while start <= end:
        mid = (start + end) // 2

        if target < nums[mid]:
            end = mid - 1

        elif target > nums[mid]:
            start = mid + 1

        else:    #nums[mid] == target
            return mid

    return -1
search([-1,2,3,6,7],6)
```

Binary Search Logic
1. Array must be sorted.
2. Find the middle element.
3. Compare middle element with target.
4. If target equals middle → found.
5. If target is greater than middle → search in right half.
6. If target is smaller than middle → search in left half.
7. Repeat steps 2–6 on the remaining half.
8. If no elements are left to search → target not found.
